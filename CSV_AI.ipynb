# 4. Agrupamiento con K-means

En esta sección se aplica K-means para identificar grupos de incidentes con características similares. No se utilizan identificadores ni columnas de texto directamente. Se seleccionan variables numéricas y se estandarizan antes de entrenar el modelo.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA

# Variables numéricas útiles para describir los incidentes
columnas_modelo = [
    "gravedad_num",
    "nivel_confianza_num",
    "dano_real_confirmado",
    "participacion_ia_confirmada",
    "source_count",
    "evento_emblematico"
]

# Agregar indicadores binarios de tipos de ataque y regiones
columnas_modelo += [c for c in df.columns if c.startswith("ataque_")]
columnas_modelo += [c for c in df.columns if c.startswith("region_")]

# Preparar matriz numérica
X = df[columnas_modelo].copy()
X = X.apply(pd.to_numeric, errors="coerce").fillna(0)

escalador = StandardScaler()
X_escalado = escalador.fit_transform(X)

print("Variables utilizadas:", len(columnas_modelo))
print("Registros utilizados:", X_escalado.shape[0])
```

### Interpretación
Se estandarizan las variables para que aquellas con valores numéricos más grandes no dominen el cálculo de distancias. K-means utiliza esas distancias para formar grupos.


## 4.1 Evaluación de diferentes valores de K

Se comparan varios valores de K mediante cuatro criterios: método del codo (inercia), coeficiente silhouette, índice Davies-Bouldin e índice Calinski-Harabasz.


```python
resultados_k = []
rango_k = range(2, 9)

for k in rango_k:
    modelo = KMeans(n_clusters=k, random_state=42, n_init=10)
    etiquetas = modelo.fit_predict(X_escalado)

    resultados_k.append({
        "K": k,
        "Inercia": modelo.inertia_,
        "Silhouette": silhouette_score(X_escalado, etiquetas),
        "Davies-Bouldin": davies_bouldin_score(X_escalado, etiquetas),
        "Calinski-Harabasz": calinski_harabasz_score(X_escalado, etiquetas)
    })

metricas_k = pd.DataFrame(resultados_k)
display(metricas_k.round(3))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

axes[0, 0].plot(metricas_k["K"], metricas_k["Inercia"], marker="o")
axes[0, 0].set_title("Método del codo")
axes[0, 0].set_xlabel("Cantidad de clusters (K)")
axes[0, 0].set_ylabel("Inercia")

axes[0, 1].plot(metricas_k["K"], metricas_k["Silhouette"], marker="o")
axes[0, 1].set_title("Coeficiente Silhouette")
axes[0, 1].set_xlabel("Cantidad de clusters (K)")
axes[0, 1].set_ylabel("Silhouette")

axes[1, 0].plot(metricas_k["K"], metricas_k["Davies-Bouldin"], marker="o")
axes[1, 0].set_title("Índice Davies-Bouldin")
axes[1, 0].set_xlabel("Cantidad de clusters (K)")
axes[1, 0].set_ylabel("Davies-Bouldin")

axes[1, 1].plot(metricas_k["K"], metricas_k["Calinski-Harabasz"], marker="o")
axes[1, 1].set_title("Índice Calinski-Harabasz")
axes[1, 1].set_xlabel("Cantidad de clusters (K)")
axes[1, 1].set_ylabel("Calinski-Harabasz")

plt.tight_layout()
plt.show()
```

### Interpretación de las métricas
- **Codo:** se busca el punto donde la reducción de la inercia empieza a ser menos pronunciada.
- **Silhouette:** valores más cercanos a 1 indican grupos más separados y compactos.
- **Davies-Bouldin:** valores más bajos indican una mejor separación entre grupos.
- **Calinski-Harabasz:** valores más altos suelen indicar grupos más compactos y separados.

Las métricas pueden sugerir cantidades distintas de clusters. La decisión también debe considerar que no se formen grupos demasiado pequeños o difíciles de interpretar.


## 4.2 Entrenamiento del modelo final

Se utiliza K=2 como primera solución de referencia: en esta ejecución ofrece el mejor valor de Silhouette y Calinski-Harabasz entre los valores evaluados. Se debe interpretar junto con las demás métricas y los tamaños de los grupos.


```python
k_final = 2

modelo_final = KMeans(n_clusters=k_final, random_state=42, n_init=10)
df["cluster"] = modelo_final.fit_predict(X_escalado)

print("Cantidad de registros por cluster:")
display(df["cluster"].value_counts().sort_index())
```

## 4.3 Perfil de los clusters

Se comparan los grupos mediante promedios de variables numéricas y proporciones de indicadores. Esto ayuda a describir qué características distinguen a cada cluster.


```python
columnas_perfil = [
    "gravedad_num",
    "nivel_confianza_num",
    "dano_real_confirmado",
    "participacion_ia_confirmada",
    "source_count",
    "evento_emblematico"
]

perfil_clusters = df.groupby("cluster")[columnas_perfil].mean().round(3)
display(perfil_clusters)
```

### Interpretación
Los promedios permiten comparar los perfiles. Por ejemplo, una proporción mayor en `dano_real_confirmado` indica que ese grupo contiene una fracción más alta de incidentes con daño confirmado. Los resultados describen patrones del conjunto de datos; no demuestran causas.


## 4.4 Visualización con PCA

PCA reduce las variables estandarizadas a dos componentes principales para representar los registros en un gráfico 2D. **PCA se utiliza aquí para visualizar los clusters**, no como una métrica adicional para elegir K.


```python
pca = PCA(n_components=2, random_state=42)
componentes = pca.fit_transform(X_escalado)

grafico_pca = pd.DataFrame({
    "Componente 1": componentes[:, 0],
    "Componente 2": componentes[:, 1],
    "Cluster": df["cluster"].astype(str).values
})

plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=grafico_pca,
    x="Componente 1",
    y="Componente 2",
    hue="Cluster",
    palette="Set2",
    alpha=0.8
)
plt.title("Visualización de clusters mediante PCA")
plt.show()

print("Varianza explicada por los dos componentes:",
      round(pca.explained_variance_ratio_.sum(), 3))
```

### Interpretación
Cada punto representa un incidente y el color identifica el cluster asignado por K-means. Los puntos cercanos tienen perfiles similares en la representación reducida. La proporción de varianza explicada indica cuánta información de las variables originales queda representada en estos dos ejes.

El análisis exploratorio permite describir los tipos de incidentes, su gravedad, las regiones y su evolución temporal. K-means complementa el EDA al agrupar registros con características similares. La solución seleccionada debe interpretarse considerando las cuatro métricas, el tamaño de los grupos y el perfil de cada cluster. PCA facilita visualizar los grupos, pero no sustituye la evaluación de las métricas.
